```python
import json
import requests

# Fetch weather data from API
url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": 14.6,
    "longitude": 121.0,
    "current": "temperature_2m"
}

response = requests.get(url, params=params)
data = response.json()

# Save to file
with open("weather_raw.json", "w") as f:
    json.dump(data, f)
```

```python
import requests
url="https://jsonplaceholder.typicode.com/posts"
response=requests.get(url)
data=response.json()
print(data[:3])
```

```python
print("Rows:", len(posts_df))
print("Columns:", posts_df.columns.tolist())

posts_df.info()
```
